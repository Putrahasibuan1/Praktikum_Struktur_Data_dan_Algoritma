list berisi minimal 20 angka dan urutkan dengan sorted()

```python
data = [45, 12, 78, 3, 56, 23, 89, 34, 67, 10,
        91, 5, 43, 72, 18, 60, 27, 84, 39, 51]

data_sorted = sorted(data)

print("Data setelah diurutkan:")
print(data_sorted)
```

```python
def binary_search_trace(data, target):
    left = 0
    right = len(data) - 1
    jumlah_pemeriksaan = 0

    while left <= right:
        mid = (left + right) // 2
        jumlah_pemeriksaan += 1

        print("left =", left, "right =", right, "mid =", mid)

        if data[mid] == target:
            return mid, jumlah_pemeriksaan

        elif data[mid] < target:
            left = mid + 1

        else:
            right = mid - 1

    return -1, jumlah_pemeriksaan
```

Jalankan binary_search_trace() untuk target yang berada di sekitar tengah data.

```python
target = 51

hasil, pemeriksaan = binary_search_trace(data_sorted, target)

print("Index:", hasil)
print("Jumlah pemeriksaan:", pemeriksaan)
```

Ulangi untuk target yang berada dekat awal dan dekat akhir data

```python
target = data_sorted[1]

hasil, pemeriksaan = binary_search_trace(data_sorted, target)

print("Target:", target)
print("Index:", hasil)
print("Jumlah pemeriksaan:", pemeriksaan)
```

```python
target = data_sorted[-2]

hasil, pemeriksaan = binary_search_trace(data_sorted, target)

print("Target:", target)
print("Index:", hasil)
print("Jumlah pemeriksaan:", pemeriksaan)
```

Catat nilai left, right, dan mid untuk salah satu pengujian dalam sebuah tabel.

```python
target = 51

left = 0
right = len(data_sorted) - 1

print("left\t right\t mid")

while left <= right:
    mid = (left + right) // 2

    print(left, "\t", right, "\t", mid)

    if data_sorted[mid] == target:
        break
    elif data_sorted[mid] < target:
        left = mid + 1
    else:
        right = mid - 1
```

Jalankan fungsi yang sama pada data yang sengaja diacak

```python
import random

data_acak = data_sorted.copy()
random.shuffle(data_acak)

print("Data acak:")
print(data_acak)

hasil, pemeriksaan = binary_search_trace(data_acak, 51)

print("Index:", hasil)
print("Jumlah pemeriksaan:", pemeriksaan)
```

Bandingkan jumlah pemeriksaan Binary Search dengan Linear Search pada target yang sama

```python
def linear_search(data, target):
    jumlah_pemeriksaan = 0

    for i in range(len(data)):
        jumlah_pemeriksaan += 1

        if data[i] == target:
            return i, jumlah_pemeriksaan

    return -1, jumlah_pemeriksaan
```
