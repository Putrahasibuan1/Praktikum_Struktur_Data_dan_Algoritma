```python
# 2 Membangun Tree dengan Class Node


class Node:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None

root = Node('K')

root.left = Node('D')
root.right = Node('M')

root.left.left = Node('B')
root.left.right = Node('F')

root.right.right = Node('Z')

print("Binary Tree berhasil dibuat!")
print("Root:", root.data)
print("Child kiri:", root.left.data)
print("Child kanan:", root.right.data)
```

```python
# 3 Menampilkan Traversal



def preorder(node):
    if node is None:
        return
    print(node.data, end=" ")
    preorder(node.left)
    preorder(node.right)

def inorder(node):
    if node is None:
        return
    inorder(node.left)
    print(node.data, end=" ")
    inorder(node.right)

def postorder(node):
    if node is None:
        return
    postorder(node.left)
    postorder(node.right)
    print(node.data, end=" ")

def level_order(node):
    if node is None:
        return

    queue = [node]

    while queue:
        current = queue.pop(0)
        print(current.data, end=" ")

        if current.left:
            queue.append(current.left)

        if current.right:
            queue.append(current.right)

print("Preorder:")
preorder(root)

print("\nInorder:")
inorder(root)

print("\nPostorder:")
postorder(root)

print("\nLevel-order:")
level_order(root)
```

```python
# 4 Menghitung Jumlah Node dan Leaf



def jumlah_node(node):
    if node is None:
        return 0

    return 1 + jumlah_node(node.left) + jumlah_node(node.right)

def jumlah_leaf(node):
    if node is None:
        return 0

    if node.left is None and node.right is None:
        return 1

    return jumlah_leaf(node.left) + jumlah_leaf(node.right)

print("Jumlah seluruh node:", jumlah_node(root))
print("Jumlah leaf:", jumlah_leaf(root))
```

```python
# 5 Menghitung Height Tree


def height_tree(node):
    if node is None:
        return 0

    tinggi_kiri = height_tree(node.left)
    tinggi_kanan = height_tree(node.right)

    return 1 + max(tinggi_kiri, tinggi_kanan)

print("Height Tree:", height_tree(root))
```

```python
# 6 Mencari Nilai F dan X dengan Fungsi Search


def search(node, nilai):
    if node is None:
        return False

    if node.data == nilai:
        return True

    return search(node.left, nilai) or search(node.right, nilai)

if search(root, 'F'):
    print("Nilai F ditemukan")
else:
    print("Nilai F tidak ditemukan")

if search(root, 'X'):
    print("Nilai X ditemukan")
else:
    print("Nilai X tidak ditemukan")
```
