7.25. Praktik Mandiri 7.1 — Bubble Sort Descending

```python
def bubble_sort(data):
    n = len(data)

    for i in range(n - 1):
        for j in range(n - 1 - i):
            if data[j] < data[j + 1]:
                data[j], data[j + 1] = data[j + 1], data[j]

    return data

nilai = [14, 4, 25, 9, 13, 2]
print(bubble_sort(nilai))
```

7.26. Praktik Mandiri 7.2 — Selection Sort dengan Statistik

```python
def selection_sort_statistik(data):
    # Salin list agar data asli tidak berubah
    arr = list(data)
    n = len(arr)

    comparisons = 0
    swaps = 0

    for i in range(n - 1):
        min_index = i
        for j in range(i + 1, n):
            comparisons += 1  # Setiap kali ada perbandingan data[j] < data[min_index]
            if arr[j] < arr[min_index]:
                min_index = j

        # Lakukan swap jika ditemukan elemen yang lebih kecil
        if min_index != i:
            arr[i], arr[min_index] = arr[min_index], arr[i]
            swaps += 1  # Setiap kali terjadi pertukaran posisi

    return arr, comparisons, swaps

# Data uji dari soal
uji_urut = [1, 2, 3, 4, 5, 6]
uji_acak = [4, 1, 6, 2, 5, 3]
uji_terbalik = [6, 5, 4, 3, 2, 1]

# Menjalankan fungsi untuk masing-masing data
for nama, data in [("Uji Urut", uji_urut), ("Uji Acak", uji_acak), ("Uji Terbalik", uji_terbalik)]:
    hasil, comp, swp = selection_sort_statistik(data)
    print(f"--- {nama} ---")
    print(f"Data awal    : {data}")
    print(f"Data akhir   : {hasil}")
    print(f"Comparisons  : {comp}")
    print(f"Swaps        : {swp}\n")
```

7.27. Praktik Mandiri 7.3 — Insertion Sort dan Data Hampir Terurut

```python
def insertion_sort_dengan_hitung(data):
    # Menyalin list agar data asli tidak berubah
    arr = list(data)
    geser = 0

    # Algoritma Insertion Sort
    for i in range(1, len(arr)):
        nilai_sekarang = arr[i]
        posisi = i

        # Pergeseran elemen ke kanan jika elemen sebelumnya lebih besar
        while posisi > 0 and arr[posisi - 1] > nilai_sekarang:
            arr[posisi] = arr[posisi - 1]
            posisi -= 1
            geser += 1  # Menghitung setiap kali pergeseran data terjadi

        arr[posisi] = nilai_sekarang

    return arr, geser

# Data dari soal
hampir = [1, 2, 3, 5, 4, 6, 7, 8]
acak = [7, 2, 8, 1, 5, 3, 6, 4]

# Menjalankan fungsi
hasil_hampir, jumlah_geser_hampir = insertion_sort_dengan_hitung(hampir)
hasil_acak, jumlah_geser_acak = insertion_sort_dengan_hitung(acak)

print(f"Data Hampir Terurut Setelah Sort: {hasil_hampir}")
print(f"Jumlah Pergeseran (Hampir Terurut): {jumlah_geser_hampir}\n")

print(f"Data Acak Setelah Sort: {hasil_acak}")
print(f"Jumlah Pergeseran (Acak): {jumlah_geser_acak}")
```
