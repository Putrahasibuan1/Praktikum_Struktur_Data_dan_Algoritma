# Praktik Mandiri 2.2 — Linear Search

Gunakan fungsi Linear Search buatan sendiri, bukan operator `in` atau method lain yang menyembunyikan proses pencarian.

Buat list minimal 15 bilangan yang tidak harus terurut

```python
data = [25, 12, 8, 40, 17, 3, 29, 11, 50, 6, 34, 21, 9, 45, 15]

print("Data:", data)
```

Membuat fungsi Linear Search

```python
def linear_search(data, target):
    jumlah_pemeriksaan = 0

    for i in range(len(data)):
        jumlah_pemeriksaan += 1

        if data[i] == target:
            return i, jumlah_pemeriksaan

    return -1, jumlah_pemeriksaan
```

Cari target yang berada pada elemen pertama

```python
target = data[0]

index, jumlah_pemeriksaan = linear_search(data, target)

print("Target:", target)
print("Index:", index)
print("Jumlah pemeriksaan:", jumlah_pemeriksaan)
```

Cari target yang berada pada elemen terakhir

```python
target = data[-1]

index, jumlah_pemeriksaan = linear_search(data, target)

print("Target:", target)
print("Index:", index)
print("Jumlah pemeriksaan:", jumlah_pemeriksaan)
```

Cari target yang tidak ada di dalam data

```python
target = 100

index, jumlah_pemeriksaan = linear_search(data, target)

print("Target:", target)
print("Index:", index)
print("Jumlah pemeriksaan:", jumlah_pemeriksaan)
```

## Perbandingan

Best case terjadi ketika target berada pada elemen pertama karena hanya membutuhkan 1 pemeriksaan. Worst case terjadi ketika target berada pada elemen terakhir atau target tidak ditemukan karena seluruh elemen harus diperiksa.

```python
print("Best Case")
print("Target:", data[0])
print("Jumlah pemeriksaan: 1")

print()

print("Worst Case - Target di elemen terakhir")
print("Target:", data[-1])
print("Jumlah pemeriksaan:", len(data))

print()

print("Worst Case - Target tidak ditemukan")
print("Target: 100")
print("Jumlah pemeriksaan:", len(data))
```

Fungsi mengembalikan pasangan (index, jumlah_pemeriksaan)

```python
hasil = linear_search(data, 29)

print("Hasil:", hasil)
```

Worst case Linear Search adalah O(n) karena pada kondisi terburuk semua elemen harus diperiksa. Kondisi ini terjadi ketika target berada pada elemen terakhir atau target tidak ditemukan. Semakin banyak jumlah data, semakin banyak pula pemeriksaan yang diperlukan. Oleh karena itu, kompleksitas waktu Linear Search pada worst case adalah O(n).

```python
def linear_search_text(data, target):
    jumlah_pemeriksaan = 0

    for i in range(len(data)):
        jumlah_pemeriksaan += 1

        if data[i].lower() == target.lower():
            return i, jumlah_pemeriksaan

    return -1, jumlah_pemeriksaan
```

```python
data_text = ["Budi", "Andi", "Siti", "Rina", "Doni"]

target = "andi"

index, jumlah_pemeriksaan = linear_search_text(data_text, target)

print("Target:", target)
print("Index:", index)
print("Jumlah pemeriksaan:", jumlah_pemeriksaan)
```
