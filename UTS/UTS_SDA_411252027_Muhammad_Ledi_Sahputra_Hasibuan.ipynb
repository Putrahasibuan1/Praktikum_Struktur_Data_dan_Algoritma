# SOAL 1 — Sistem Pencarian Data Mahasiswa dan Pengurutan Rekursif

## A. Tujuan
Membuat program untuk mencari data mahasiswa berdasarkan NIM. Data diurutkan secara ascending menggunakan **Merge Sort rekursif**, kemudian pencarian dilakukan menggunakan **Linear Search** dan **Binary Search**.

## B. Parameter dan data awal
Digit terakhir NIM adalah **d = 7**. Oleh karena itu:
- Nilai Bima = 80 + 7 = 87
- Nilai Hadi = 65 + 7 = 72
- Nilai Lukman = 70 + 7 = 77

Target indeks ditentukan dengan rumus `d % len(data)`, yaitu `7 % 12 = 7`. Dengan indeks Python yang dimulai dari 0, target NIM adalah data pada indeks ke-7.


```python
# SOAL 1 — Data awal
d = 7

data = [
    ['TI001', 'Alya', 78],
    ['TI002', 'Bima', 80 + d],
    ['TI003', 'Citra', 67],
    ['TI004', 'Danu', 92],
    ['TI005', 'Eka', 74],
    ['TI006', 'Fajar', 88],
    ['TI007', 'Gina', 81],
    ['TI008', 'Hadi', 65 + d],
    ['TI009', 'Intan', 91],
    ['TI010', 'Joko', 76],
    ['TI011', 'Karin', 84],
    ['TI012', 'Lukman', 70 + d]
]

target_index = d % len(data)
target_nim = data[target_index][0]

print("Digit terakhir NIM (d):", d)
print("Jumlah data:", len(data))
print("Target index:", target_index)
print("Target NIM:", target_nim)
print("\nData awal:")
for mahasiswa in data:
    print(mahasiswa)
```

## C. Implementasi Linear Search manual

Linear Search memeriksa kode mahasiswa satu per satu dari awal sampai target ditemukan atau seluruh data selesai diperiksa. Fungsi mengembalikan indeks data dan jumlah perbandingan. Jika target tidak ditemukan, indeks bernilai `-1`.


```python
def linear_search_nim(data, target):
    perbandingan = 0

    for i in range(len(data)):
        perbandingan += 1
        if data[i][0] == target:
            return i, perbandingan

    return -1, perbandingan


# Pengujian Linear Search: target ditemukan dan tidak ditemukan
indeks_linear, banding_linear = linear_search_nim(data, target_nim)
indeks_tidak_ada, banding_tidak_ada = linear_search_nim(data, 'TI999')

print("Linear Search untuk", target_nim)
print("Indeks hasil:", indeks_linear)
print("Jumlah perbandingan:", banding_linear)

print("\nLinear Search untuk TI999")
print("Indeks hasil:", indeks_tidak_ada)
print("Jumlah perbandingan:", banding_tidak_ada)
```

## D. Implementasi Merge Sort rekursif manual

Merge Sort membagi data menjadi dua bagian secara rekursif sampai setiap bagian berisi satu elemen. Setelah itu, bagian-bagian tersebut digabung kembali dalam urutan kode mahasiswa ascending.

Implementasi berikut tidak menggunakan `sorted()`, `list.sort()`, `bisect`, atau library pengurutan.


```python
def merge(left, right):
    hasil = []
    i = 0
    j = 0

    # Menggabungkan dua bagian yang sudah terurut
    while i < len(left) and j < len(right):
        if left[i][0] <= right[j][0]:
            hasil.append(left[i])
            i += 1
        else:
            hasil.append(right[j])
            j += 1

    # Menambahkan sisa elemen
    while i < len(left):
        hasil.append(left[i])
        i += 1

    while j < len(right):
        hasil.append(right[j])
        j += 1

    return hasil


def merge_sort(data, tampilkan_trace=False, tingkat=0):
    # Kondisi dasar rekursi
    if len(data) <= 1:
        return data[:]

    tengah = len(data) // 2
    kiri = data[:tengah]
    kanan = data[tengah:]

    if tampilkan_trace:
        indentasi = "  " * tingkat
        print(indentasi + "Bagi:", [item[0] for item in data])

    # Pemanggilan rekursif
    kiri_urut = merge_sort(kiri, tampilkan_trace, tingkat + 1)
    kanan_urut = merge_sort(kanan, tampilkan_trace, tingkat + 1)

    # Proses penggabungan
    hasil = merge(kiri_urut, kanan_urut)

    if tampilkan_trace:
        indentasi = "  " * tingkat
        print(indentasi + "Gabung:", [item[0] for item in hasil])

    return hasil


# Trace singkat Merge Sort pada 6 data pertama
print("Trace Merge Sort untuk 6 data pertama:")
contoh_enam_data = [baris[:] for baris in data[:6]]
hasil_trace = merge_sort(contoh_enam_data, tampilkan_trace=True)

# Mengurutkan seluruh data awal tanpa mengubah data asli
data_terurut = merge_sort([baris[:] for baris in data])

print("\nData sebelum pengurutan:")
for mahasiswa in data:
    print(mahasiswa)

print("\nData setelah Merge Sort (ascending):")
for mahasiswa in data_terurut:
    print(mahasiswa)
```

## E. Implementasi Binary Search manual

Binary Search bekerja pada data yang sudah terurut. Pencarian dilakukan dengan memeriksa elemen tengah, lalu mempersempit batas pencarian ke bagian kiri atau kanan.

Jumlah perbandingan menghitung setiap pemeriksaan kesamaan (`==`) dan pemeriksaan lebih kecil (`<`) terhadap kode di posisi tengah.


```python
def binary_search_nim(data_urut, target):
    kiri = 0
    kanan = len(data_urut) - 1
    perbandingan = 0

    while kiri <= kanan:
        tengah = (kiri + kanan) // 2

        perbandingan += 1
        if data_urut[tengah][0] == target:
            return tengah, perbandingan

        perbandingan += 1
        if target < data_urut[tengah][0]:
            kanan = tengah - 1
        else:
            kiri = tengah + 1

    return -1, perbandingan


# Pengujian Binary Search pada data yang sudah terurut
indeks_binary, banding_binary = binary_search_nim(data_terurut, target_nim)
indeks_binary_tidak_ada, banding_binary_tidak_ada = binary_search_nim(data_terurut, 'TI999')

print("Binary Search untuk", target_nim)
print("Indeks hasil pada data terurut:", indeks_binary)
print("Jumlah perbandingan:", banding_binary)

print("\nBinary Search untuk TI999")
print("Indeks hasil:", indeks_binary_tidak_ada)
print("Jumlah perbandingan:", banding_binary_tidak_ada)
```

## G. Analisis kompleksitas Soal 1

1. **Linear Search**
   - Waktu terbaik: `O(1)`, jika target berada pada elemen pertama.
   - Waktu rata-rata dan terburuk: `O(n)`, karena elemen dapat diperiksa satu per satu.
   - Ruang tambahan: `O(1)` karena hanya menggunakan beberapa variabel.

2. **Binary Search**
   - Waktu terbaik: `O(1)`, jika target langsung berada di posisi tengah.
   - Waktu rata-rata dan terburuk: `O(log n)`, karena ruang pencarian dibagi dua pada setiap iterasi.
   - Ruang tambahan: `O(1)` karena implementasinya iteratif.
   - Syarat: data harus sudah terurut berdasarkan kode mahasiswa.

3. **Merge Sort**
   - Waktu terbaik, rata-rata, dan terburuk: `O(n log n)`.
   - Ruang tambahan: `O(n)` untuk proses penggabungan, serta ruang pemanggilan rekursif sekitar `O(log n)`.
   - Merge Sort digunakan karena pengurutan dilakukan secara rekursif dan hasilnya stabil untuk elemen dengan kunci yang sama.

4. **Kapan biaya sorting layak dibayar?**  
   Jika hanya melakukan satu pencarian, Linear Search dapat lebih hemat karena tidak perlu mengurutkan data terlebih dahulu. Jika data yang sama akan dicari berkali-kali, biaya Merge Sort dapat menjadi layak karena pencarian berikutnya menggunakan Binary Search yang lebih cepat. Secara umum, perbandingan biaya dapat dilihat dari `O(n log n + k log n)` untuk sorting sekali dan k pencarian Binary Search, dibandingkan `O(k n)` untuk k pencarian Linear Search.


# SOAL 2 — Sistem Antrean Layanan dengan Undo Menggunakan Linked Queue dan Stack

## A. Tujuan
Membuat simulasi antrean layanan mahasiswa berdasarkan prinsip FIFO (First In, First Out). Queue dibuat menggunakan **singly linked list** dengan referensi `front`, `rear`, dan `count`. Stack dibuat menggunakan Python list sebagai penyimpanan internal, dengan operasi push dan pop yang dibungkus di dalam class Stack.

Ketentuan implementasi:
- Queue tidak menggunakan `collections.deque`, `queue.Queue`, atau `list.pop(0)`.
- Queue menggunakan class `Node` dan `LinkedQueue`.
- Stack menggunakan Python list internal dan class `Stack`.
- Operasi `UNDO` mengambil ID terakhir dari Stack dan memasukkannya kembali ke bagian depan Queue.


## B. Implementasi Node dan LinkedQueue

Method yang dibuat: `enqueue(data)`, `enqueue_front(data)`, `dequeue()`, `peek()`, `is_empty()`, `size()`, dan `display()`.


```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


class LinkedQueue:
    def __init__(self):
        self.front = None
        self.rear = None
        self.count = 0

    def enqueue(self, data):
        node_baru = Node(data)

        if self.is_empty():
            self.front = node_baru
            self.rear = node_baru
        else:
            self.rear.next = node_baru
            self.rear = node_baru

        self.count += 1

    def enqueue_front(self, data):
        node_baru = Node(data)

        if self.is_empty():
            self.front = node_baru
            self.rear = node_baru
        else:
            node_baru.next = self.front
            self.front = node_baru

        self.count += 1

    def dequeue(self):
        if self.is_empty():
            return None

        data_keluar = self.front.data
        self.front = self.front.next
        self.count -= 1

        if self.front is None:
            self.rear = None

        return data_keluar

    def peek(self):
        if self.is_empty():
            return None
        return self.front.data

    def is_empty(self):
        return self.front is None

    def size(self):
        return self.count

    def display(self):
        elemen = []
        current = self.front

        while current is not None:
            elemen.append(current.data)
            current = current.next

        return elemen
```

## C. Implementasi Stack

Stack menggunakan Python list sebagai penyimpanan internal. Elemen paling akhir pada list merupakan bagian atas (top) Stack. Method yang dibuat adalah `push(data)`, `pop()`, `peek()`, `is_empty()`, `size()`, dan `display()`.


```python
class Stack:
    def __init__(self):
        self.items = []

    def push(self, data):
        self.items.append(data)

    def pop(self):
        if self.is_empty():
            return None
        return self.items.pop()

    def peek(self):
        if self.is_empty():
            return None
        return self.items[-1]

    def is_empty(self):
        return len(self.items) == 0

    def size(self):
        return len(self.items)

    def display(self):
        # Urutan tampilan dari bawah ke atas Stack
        return self.items[:]
```

## D. Pengujian dasar class dan penanganan edge case

Pengujian dilakukan sebelum simulasi utama. Queue kosong harus aman ketika `dequeue()` dipanggil, dan Stack kosong harus aman ketika `pop()` dipanggil.


```python
# Pengujian dasar LinkedQueue
uji_queue = LinkedQueue()
print("Queue kosong:", uji_queue.is_empty())
print("Dequeue pada Queue kosong:", uji_queue.dequeue())
uji_queue.enqueue("UJI-A")
uji_queue.enqueue("UJI-B")
print("Isi Queue setelah enqueue:", uji_queue.display())
print("Peek Queue:", uji_queue.peek())
print("Dequeue:", uji_queue.dequeue())
print("Isi Queue sekarang:", uji_queue.display())
print("Ukuran Queue:", uji_queue.size())

# Pengujian dasar Stack
uji_stack = Stack()
print("\nStack kosong:", uji_stack.is_empty())
print("Pop pada Stack kosong:", uji_stack.pop())
uji_stack.push("UJI-A")
uji_stack.push("UJI-B")
print("Isi Stack (bawah ke atas):", uji_stack.display())
print("Peek Stack:", uji_stack.peek())
print("Pop:", uji_stack.pop())
print("Isi Stack sekarang:", uji_stack.display())
print("Ukuran Stack:", uji_stack.size())
```

## E. Simulasi 12 operasi sesuai soal

Dengan `d = 7`, ID layanan yang digunakan adalah:
- A = `MHS-701`
- B = `MHS-702`
- C = `MHS-703`
- D = `MHS-704`
- E = `MHS-705`
- F = `MHS-706`

Arti operasi:
- **DATANG X:** memasukkan ID ke belakang Queue.
- **LAYANI:** mengambil ID dari depan Queue, lalu menyimpan ID tersebut ke Stack riwayat jika berhasil.
- **UNDO:** mengambil satu ID terakhir dari Stack, kemudian mengembalikannya ke bagian depan Queue.

Setelah setiap operasi, isi Queue dan Stack ditampilkan.


```python
# Membentuk ID layanan berdasarkan digit terakhir NIM
d = 7
id_layanan = {
    "A": f"MHS-{d}01",
    "B": f"MHS-{d}02",
    "C": f"MHS-{d}03",
    "D": f"MHS-{d}04",
    "E": f"MHS-{d}05",
    "F": f"MHS-{d}06"
}

queue = LinkedQueue()
riwayat = Stack()

operasi = [
    ("DATANG", "A"),
    ("DATANG", "B"),
    ("DATANG", "C"),
    ("LAYANI", None),
    ("DATANG", "D"),
    ("LAYANI", None),
    ("DATANG", "E"),
    ("UNDO", None),
    ("LAYANI", None),
    ("LAYANI", None),
    ("UNDO", None),
    ("DATANG", "F")
]

def tampilkan_status(langkah, nama_operasi, keterangan):
    print(f"\nLangkah {langkah}: {nama_operasi} — {keterangan}")
    print("Queue (front -> rear):", queue.display())
    print("Stack (bawah -> atas):", riwayat.display())
    print("Jumlah Queue:", queue.size(), "| Jumlah Stack:", riwayat.size())


for nomor, (jenis, kode) in enumerate(operasi, start=1):
    if jenis == "DATANG":
        identitas = id_layanan[kode]
        queue.enqueue(identitas)
        keterangan = f"Mahasiswa {identitas} masuk ke belakang antrean"

    elif jenis == "LAYANI":
        identitas = queue.dequeue()

        if identitas is None:
            keterangan = "Queue kosong, tidak ada mahasiswa yang dapat dilayani"
        else:
            riwayat.push(identitas)
            keterangan = f"{identitas} dilayani dan disimpan ke Stack"

    elif jenis == "UNDO":
        identitas = riwayat.pop()

        if identitas is None:
            keterangan = "Stack kosong, tidak ada layanan yang dapat di-undo"
        else:
            queue.enqueue_front(identitas)
            keterangan = f"{identitas} dikembalikan ke depan Queue"

    tampilkan_status(nomor, jenis if kode is None else f"{jenis} {kode}", keterangan)
```

## F. Ringkasan kondisi akhir Soal 2

Setelah seluruh 12 operasi selesai, program menampilkan elemen Queue dari front ke rear dan elemen Stack dari bawah ke atas. Stack menyimpan riwayat layanan yang belum dibatalkan oleh operasi Undo.


```python
print("========== RINGKASAN SOAL 2 ==========")
print("Isi akhir Queue (front -> rear):", queue.display())
print("Front Queue:", queue.peek())
print("Rear Queue:", queue.rear.data if queue.rear is not None else None)
print("Jumlah elemen Queue:", queue.size())
print("Isi akhir Stack (bawah -> atas):", riwayat.display())
print("Top Stack:", riwayat.peek())
print("Jumlah elemen Stack:", riwayat.size())
```

## H. Analisis FIFO, LIFO, dan kompleksitas Soal 2

### 1. FIFO pada Queue
Queue mengikuti prinsip **FIFO (First In, First Out)**. Mahasiswa yang datang lebih dahulu akan berada di depan antrean dan dilayani lebih dahulu. Contohnya, ketika A, B, dan C datang berurutan, A akan dilayani sebelum B dan C.

### 2. LIFO pada Stack
Stack mengikuti prinsip **LIFO (Last In, First Out)**. Layanan yang paling terakhir disimpan ke Stack akan menjadi layanan pertama yang dibatalkan ketika Undo dilakukan. Contohnya, jika riwayat Stack berisi A lalu B, operasi Undo akan mengambil B terlebih dahulu.

### 3. Kompleksitas operasi LinkedQueue
- `enqueue(data)`: `O(1)` karena elemen baru ditambahkan melalui referensi `rear`.
- `enqueue_front(data)`: `O(1)` karena elemen baru langsung menjadi `front`.
- `dequeue()`: `O(1)` karena hanya memindahkan referensi `front`.
- `peek()`: `O(1)` karena langsung membaca elemen `front`.
- `is_empty()`: `O(1)` karena memeriksa referensi `front`.
- `size()`: `O(1)` karena jumlah elemen disimpan pada atribut `count`.
- `display()`: `O(n)` karena harus menelusuri seluruh node.

### 4. Kompleksitas operasi Stack
- `push(data)`: `O(1)` amortisasi karena menambahkan elemen ke akhir list.
- `pop()`: `O(1)` amortisasi karena mengambil elemen terakhir list.
- `peek()`: `O(1)` karena membaca elemen terakhir.
- `is_empty()`: `O(1)` karena memeriksa panjang list.
- `size()`: `O(1)` karena membaca panjang list.
- `display()`: `O(n)` karena menyalin seluruh isi Stack.

### 5. Mengapa Queue menggunakan Linked List?
Linked list lebih tepat untuk antrean yang sering melakukan `dequeue()` karena elemen depan dapat dilepas dengan memindahkan referensi `front` dalam `O(1)`. Jika menggunakan list Python dan `pop(0)`, elemen-elemen lain harus bergeser sehingga membutuhkan `O(n)` waktu. Dengan referensi `front` dan `rear`, operasi utama antrean dapat dilakukan secara efisien.
